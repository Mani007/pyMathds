```python
# Tensor creation
print("arrange: ", torch.arange(0,10,2)) # 0 to 10 with steps of 2 
print("Linear space: ", torch.linspace(0,20,10)) # 0 to 20 in 10 division 
print("EYE : ", torch.eye(5)) # 1 in diagonal for 5x5 tensor  
print("Full : ", torch.full((3,3),7)) # 3x3 tensor with 7 in all values  
```
