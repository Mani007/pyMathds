```python
# implement linear regression from scratch
import numpy as np
import matplotlib.pyplot as plt

# Sample data
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

# Calculate means
x_mean = np.mean(x)
y_mean = np.mean(y)

# Calculate slope (m)
numerator = np.sum((x - x_mean) * (y - y_mean))
denominator = np.sum((x - x_mean) ** 2)

m = numerator / denominator

# Calculate intercept (b)
b = y_mean - m * x_mean

print("Slope (m):", m)
print("Intercept (b):", b)

# Predict values
y_pred = m * x + b

# Predict for a new value
x_new = 6
y_new = m * x_new + b
print(f"Prediction for x={x_new}: {y_new}")

# Plot
plt.scatter(x, y, color='blue', label='Actual Data')
plt.plot(x, y_pred, color='red', label='Regression Line')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Simple Linear Regression From Scratch")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Sample data
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 5, 4, 5], dtype=float)

# Parameters (initialize)
m = 0   # slope
b = 0   # intercept

# Hyperparameters
learning_rate = 0.01
epochs = 1000
n = len(x)

# Gradient Descent
for i in range(epochs):

    # Predictions
    y_pred = m * x + b

    # Compute gradients
    dm = (-2 / n) * np.sum(x * (y - y_pred))
    db = (-2 / n) * np.sum(y - y_pred)

    # Update parameters
    m = m - learning_rate * dm
    b = b - learning_rate * db

    # Print loss every 100 epochs
    if i % 100 == 0:
        loss = np.mean((y - y_pred) ** 2)
        print(f"Epoch {i}, Loss: {loss:.4f}")

print("\nFinal Parameters:")
print("Slope (m):", m)
print("Intercept (b):", b)

# Final predictions
y_pred = m * x + b

# Predict new value
x_new = 6
y_new = m * x_new + b
print(f"Prediction for x={x_new}: {y_new}")

# Plot results
plt.scatter(x, y, color='blue', label='Actual Data')
plt.plot(x, y_pred, color='red', label='Regression Line')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression using Gradient Descent")
plt.legend()
plt.show()
```

```python
import numpy as np

def gradient_descent(x, y, learning_rate=0.01, epochs=1000):
    # Initialize parameters
    m = 0
    b = 0

    n = len(x)

    # Training loop
    for i in range(epochs):

        # Predicted values
        y_pred = m * x + b

        # Compute gradients
        dm = (-2 / n) * np.sum(x * (y - y_pred))
        db = (-2 / n) * np.sum(y - y_pred)

        # Update parameters
        m = m - learning_rate * dm
        b = b - learning_rate * db

    return m, b


# Example usage
x = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([2, 4, 5, 4, 5], dtype=float)

m, b = gradient_descent(x, y)

print("Slope:", m)
print("Intercept:", b)

# Prediction
x_new = 6
y_new = m * x_new + b
print("Prediction for x=6:", y_new)
```

```python
# Simple Gradient Descent for Linear Regression (No Libraries)

def gradient_descent(x, y, learning_rate=0.01, epochs=1000):

    # Initialize parameters
    m = 0
    b = 0

    n = len(x)

    # Training loop
    for epoch in range(epochs):

        dm = 0
        db = 0

        # Calculate gradients
        for i in range(n):

            y_pred = m * x[i] + b
            error = y[i] - y_pred

            dm += (-2 / n) * x[i] * error
            db += (-2 / n) * error

        # Update parameters
        m = m - learning_rate * dm
        b = b - learning_rate * db

    return m, b


# Example data
x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]

# Train model
m, b = gradient_descent(x, y)

print("Slope (m):", m)
print("Intercept (b):", b)

# Prediction
x_new = 6
y_new = m * x_new + b

print("Prediction for x=6:", y_new)
```

```python
def mse_loss(x, y, m, b):

    n = len(x)
    total_error = 0

    for i in range(n):

        y_pred = m * x[i] + b
        error = y[i] - y_pred

        total_error += error ** 2

    return total_error / n


# Example
x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]

m = 0.6
b = 2.2

loss = mse_loss(x, y, m, b)

print("Loss:", loss)
```
