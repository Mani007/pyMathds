```python
# Custom tensors
my_tensor = torch.tensor([[1,2,3],[4,5,6]])
my_tensor
```

```python
# Tensor creation
print("arrange: ", torch.arange(0,10,2)) # 0 to 10 with steps of 2 
print("Linear space: ", torch.linspace(0,20,10)) # 0 to 20 in 10 division 
print("EYE : ", torch.eye(5)) # 1 in diagonal for 5x5 tensor  
print("Full : ", torch.full((3,3),7)) # 3x3 tensor with 7 in all values  
```

```python
# my_tensor
my_tensor.shape # two rows and three column 
```

```python
# creating similar shape tensor as my_tensor with random data
my_tensor1 = torch.empty_like(my_tensor)
my_tensor1 
```

```python
# creating similar shape tensor as my_tensor with all zero
my_tensor0 = torch.zeros_like(my_tensor)
my_tensor0
```

```python
# creating similar shape tensor as my_tensor with all ones
my_tensor11 = torch.ones_like(my_tensor)
my_tensor11
```

```python
# creating similar shape tensor as my_tensor with all random float value
my_tensor2 = torch.rand_like(my_tensor)
my_tensor2
```
