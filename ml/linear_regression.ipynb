```python
# building an error calculator
def linear_error_calc(input_stream, output_stream):
    # check the shape of inputs
    if input_stream.shape == output_stream.shape:
        # create a new array with output - input
        return np.array(output_stream - input_stream)
    else:
        return "The shape of input and output mismatch" 
```

```python
error_linear = linear_error_calc(in_data_size,out_data_price)
error_linear
```

```python
# building an error calculator
def linear_square_error_calc(input_stream, output_stream):
    # check the shape of inputs
    if input_stream.shape == output_stream.shape:
        # create a new array with (output - input)**2
        return np.array((output_stream - input_stream)**2)
    else:
        return "The shape of input and output mismatch" 
```

```python
square_error_linear = linear_square_error_calc(in_data_size,out_data_price)
square_error_linear
```
